```python
import rustworkx as rx
import rustworkx.generators as rx_gen
from rustworkx.visualization import mpl_draw
import matplotlib.pyplot as plt
d = 5

graph = rx_gen.heavy_hex_graph(d, multigraph=False)

plt.figure(figsize=(8, 8))
mpl_draw(
    graph, 
    with_labels=True, 
    node_color="lightblue", 
    node_size=300, 
    edge_color="gray"
)

print(rx.graph_unweighted_average_shortest_path_length(graph))

plt.title(f"Heavy Hex Graph (Distance {d}) via Rustworkx")
plt.show()
```

```python
square_graph = rx_gen.grid_graph(7,7)

plt.figure(figsize=(8, 8))
mpl_draw(
    square_graph, 
    with_labels=True, 
    node_color="lightblue", 
    node_size=300, 
    edge_color="gray"
)

print(rx.graph_unweighted_average_shortest_path_length(square_graph))

plt.title(f"Heavy Hex Graph (Distance {d}) via Rustworkx")
plt.show()
```
