```python
import rustworkx as rx
import rustworkx.generators as rx_gen
from rustworkx.visualization import mpl_draw
import matplotlib.pyplot as plt
d = 7

graph = rx_gen.heavy_hex_graph(d, multigraph=False)

plt.figure(figsize=(8, 8))
mpl_draw(
    graph, 
    with_labels=True, 
    node_color="lightblue", 
    node_size=300, 
    edge_color="gray"
)

print(rx.graph_unweighted_average_shortest_path_length(graph))
n = int((5*d**2-2*d-1)/2)
plt.title(f"Number of nodes={n}")
plt.show()
```

```python
nx = 1
ny = 1
square_graph = rx_gen.grid_graph(nx,ny)
plt.figure()
mpl_draw(
    square_graph, 
    with_labels=True, 
    node_color="lightblue", 
    node_size=300, 
    edge_color="gray"
)

print(rx.graph_unweighted_average_shortest_path_length(square_graph))

plt.title(f"Number of nodes={nx*ny}")
plt.show()
```
